```python
import os
```

```python
os.makedirs('resultados/filtros', exist_ok=True)
```

```python
filtros = {
    'vendas_concluidas': dados.query('status == "concluida"'),
    'vendas_canceladas': dados.query('status == "cancelada"'),
    'vendas_pendentes': dados.query('status == "pendente"'),
    'vendas_acima_1000': dados.query('faturamento > 1000'),
    'vendas_pix': dados.query('forma_pagamento == "pix"'),
    'vendas_cartao': dados.query('forma_pagamento == "cartao"'),
    'vendas_boleto': dados.query('forma_pagamento == "boleto"'),
    'vendas_rs': dados.query('estado == "RS"')
}
```

```python
for nome_arquivo, filtro in filtros.items():
    filtro.to_csv(f'resultados/filtros/{nome_arquivo}.csv', index=False)
```
